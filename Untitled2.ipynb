```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn import tree
from sklearn.metrics import accuracy_score,confusion_matrix
from sklearn.metrics import precision_score, recall_score, f1_score

import pandas as pd
import numpy as np
import graphviz
```

```python
cancer_data=load_breast_cancer(as_frame=True)
```

```python
#get features and names
x=cancer_data['data']
y=cancer_data['target']
feature_names=cancer_data['feature_names']
target_names=cancer_data['target_names']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=0)
clf=tree.DecisionTreeClassifier()
#train
clf=clf.fit(x_train,y_train)
#train error
y_predict_train=clf.predict(x_train)
#test error
y_predict_test=clf.predict(x_test)
print("depth od the tress",clf.get_depth())
print("no of leaves of the decision tree",clf.get_n_leaves())
#accuracy
print("accuracy on train data",accuracy_score(y_train,y_predict_train))
print("accuracy on test data",accuracy_score(y_test,y_predict_test))
precision = precision_score(y_test, y_predict_test)
recall = recall_score(y_test, y_predict_test)
f1 = f1_score(y_test, y_predict_test)

print("Precision:", precision)
print("Recall:", recall)
print("F1-score:", f1)
```

graphviz

```python
dot_data = tree.export_graphviz(
    clf,
    feature_names=feature_names,
    class_names=target_names,
    filled=True
)
graph = graphviz.Source(dot_data)
graph.render("default_params")
from IPython.display import Image
graph.render("default_params", format='png')
Image(filename='default_params.png')
```
