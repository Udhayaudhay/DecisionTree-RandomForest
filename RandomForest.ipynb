IDENTIFYING  IMPORTANT FEATURES IN RANDOM FEATURES

```python
#load libraries
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn import datasets
```

```python
#load data
iris=datasets.load_iris()
feature=iris.data
target=iris.target
```

```python
#create a random  forest classifier
rd=RandomForestClassifier(random_state=0,n_jobs=-1)
```

```python
model=rd.fit(feature,target)
```

```python
#calculate  feature important
importances=model.feature_importances_
```

```python
importances
```

```python
indices=np.argsort(importances)[::-1]
```

```python
indices
```

```python
iris.feature_names
```

```python
name=[iris.feature_names[i] for i in indices]
```

```python
name
```

```python
feature.shape[1]
```

```python
importances[indices]
```

```python
plt.figure
plt.title('feature importance')
plt.bar(range(feature.shape[1]),importances[indices])
plt.xticks(range(feature.shape[1]),name,rotation=90)
plt.show()
```
